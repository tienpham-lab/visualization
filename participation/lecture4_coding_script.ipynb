```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
fig, (ax1, ax2) = plt.subplots (ncols=2, nrows=1, figsize=(7,3))
ax1.scatter(x1,y1,color='tomato',label = "Study 1")
ax2.bar(x2,y2,color='teal', label="Study 2")
ax1.legend(loc='lower left')
ax2.legend(loc='lower left')
fig.show()
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], ['ax2', 'ax3']], figsize=(7, 4))
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
plt.show()
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], ['ax2', 'ax3']], figsize=(7, 4), layout="constrained")
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label',fontsize=15,color='tomato')
someaxes["ax2"].set_xlabel('Another Label',fontsize=15, color='tomato')
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=15, color='tomato')
```

```python
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])
fig, ax = plt.subplots(figsize=(7, 3))
ax.bar(x, y1, color = "teal")
ax.plot(x, y2,color = "tomato")
```

```python
y2_sd = np.std(y2)
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "tomato")
ax.errorbar(x,y2,
yerr = y2_sd,
fmt = "none",
ecolor= "indigo",
elinewidth= 2,
capsize = 4,
capthick= 2)
```

```python
from PIL import Image
import requests
from io import BytesIO

url = "https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png"

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "image/*,*/*;q=0.8",
    "Referer": "https://en.wikipedia.org/"
}

response=requests.get(url,headers=headers)
image_file=BytesIO(response.content)
image=Image.open(image_file)
image
```

```python
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "tomato")
ax_image = fig.add_axes([0.1,0.11,0.15,0.35])
ax_image.imshow(image)
ax_image.axis('off')
filename = 'fig1a.png'
plt.savefig(filename, dpi=300)
```
